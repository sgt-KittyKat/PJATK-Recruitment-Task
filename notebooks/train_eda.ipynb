```python
## Test csv data creates issues with parsing, so it needs to be handeled manually. Judging by the error message (function reads more fields in some of the records), the problem is that some of the fields contain commas and semicolons, which are not properly escaped.
def read_test_csv(path):
    cols = ['id', 'seller_id', 'title', 'description', 'declared_category', 
            'brand', 'condition', 'price_pln', 'shipping_weight_kg', 'pack_size']
    rows, bad = [], []
    
    with open(path, encoding='utf-8') as f:
        lines = f.read().splitlines()[1:]
        
    for i, raw in enumerate(lines, start=2):
        line = raw.rstrip(';')
        if line.startswith('"') and line.endswith('"'):
            line = line[1:-1]
            
        parts = line.rsplit(',', 6)
        if len(parts) != 7:
            bad.append(i)
            continue
            
        head, *tail = parts
        hparts = head.split(',', 3)
        
        if len(hparts) < 3:
            bad.append(i)
            continue
            
        id_, seller, title = hparts[:3]
        desc = hparts[3].replace('""', '"').strip('"') if len(hparts) == 4 else ''
        
        rows.append([id_, seller, title.strip('"'), desc, *tail])
        
    if bad:
        warnings.warn(f"Skipped {len(bad)} unrecoverable lines. First 5: {bad[:5]}")
        
    df = pd.DataFrame(rows, columns=cols)
    num_cols = ['price_pln', 'shipping_weight_kg', 'pack_size']
    df[num_cols] = df[num_cols].apply(pd.to_numeric, errors='coerce')
    
    return df
```

```python
# counts and rates together
stats = train.groupby("declared_category")["target"].agg(
    count="count",
    positives="sum",
    corruption_rate="mean"
).sort_values(by="corruption_rate", ascending=False)

print(stats)

# Plot with 95% Wilson confidence intervals
import statsmodels.stats.proportion as smp

ci = smp.proportion_confint(stats["positives"], stats["count"], method="wilson")
stats["ci_lower"] = ci[0]
stats["ci_upper"] = ci[1]
stats["err_lower"] = stats["corruption_rate"] - stats["ci_lower"]
stats["err_upper"] = stats["ci_upper"] - stats["corruption_rate"]

stats["corruption_rate"].plot(
    kind="bar",
    yerr=[stats["err_lower"], stats["err_upper"]],
    capsize=4,
    figsize=(10, 5),
    title="Corruption Rate with 95% Confidence Intervals"
)
```

```python
train[train["declared_category"] == "music"].value_counts("target").plot(kind = 'bar', title = "Controlled data corruption in music category")
```

Music category has the biggest corruption rate, however it also represents the smallest sample size (Law of small numbers could be at play here). Therefore we also showcased the interval levels. 
We need to investigate this category further since small categories with extreme rates will cause overfitting in a predictive model. 

```python
music_mask = train["declared_category"] == "music"
music_corr = train.loc[music_mask, "target"].sum()
music_uncorr = len(train.loc[music_mask]) - music_corr

rest_corr = train.loc[~music_mask, "target"].sum()
rest_uncorr = len(train.loc[~music_mask]) - rest_corr

odds_ratio, p_value = fisher_exact([[music_corr, music_uncorr], 
                                    [rest_corr, rest_uncorr]])

print(f"p-value: {p_value:.4f}")
```

P value is < 0,05 therefore we can reject the null hypothesis that music category rate is the same as the rest of the dataset. 
The differenceis statisticaly significant. 

```python
# Chi-square test of independence across ALL categories simultaneously
contingency_table = pd.crosstab(train["declared_category"], train["target"])
chi2, p_val, dof, expected = stats.chi2_contingency(contingency_table)

print(f"Overall dataset p-value across all categories: {p_val}")
```

The P value for the Chi-square test of independence for all categories is again < 0,05 which means that "declared_category" is statistically significant, definitely tied to the corruption injection process and is a valid feature to feed into the pipeline.

```python

table = pd.crosstab(train["declared_category"], train["target"])

chi2, p_val, dof, expected = chi2_contingency(table)
expected_df = pd.DataFrame(expected, index=table.index, columns=table.columns)


residuals = (table - expected_df) / np.sqrt(expected_df)


corruption_analysis = pd.DataFrame({
    "Observed_Corrupted": table[1],
    "Expected_Corrupted": expected_df[1].round(1),
    "Total_Items": table.sum(axis=1),
    "Observed_Rate": (table[1] / table.sum(axis=1)).round(4),
    "Z_Score_Residual": residuals[1].round(2)
}).sort_values(by="Z_Score_Residual", ascending=False)

print(corruption_analysis)
```

A Score residual for the music and garden categories suggests that these categories were significantly targeted for corruption. While sport and books categories were significantly deprioritized from corruption.
